```python
import pandas as pd
import numpy as np

customers = pd.read_csv("../../data/raw/customers.csv")

customers.info()
```

```python
order_items = pd.read_csv("../../data/raw/order_items.csv")

order_items.info()
```

```python
orders = pd.read_csv("../../data/raw/orders.csv")

orders.info()
```

```python
products = pd.read_csv("../../data/raw/products.csv")

products.info()
```

```python
# 날짜 타입 / 결측치 확인
orders["order_date"] = pd.to_datetime(orders["order_date"],errors="coerce")
orders["order_date"].info()
```

```python
orders.head()
```

```python
# 키 결측 확인(부모 테이블 중 결측 확인)
print(customers["customer_id"].isna().sum())
print(order_items["order_item_id"].isna().sum())
print(orders["order_id"].isna().sum())
print(products["product_id"].isna().sum())

```

```python
# 파생컬럼 total_price를 order_items에 추가

order_items["total_price"] = order_items["quantity"]*order_items["unit_price"]
order_items.head()
```

```python
# 상품 카테고리의 개수와 가격 통계
product_count = products["category"].value_counts(dropna=False).to_frame()
type(product_count)
```

```python
# 상품 카테고리별 분포
category_price = products.groupby("category",dropna=False).agg(
    product_count = ("product_id", "size"),
    mean_price = ("price","mean"),
    median_price = ("price","median")
)
print(category_price)
```

완료 주문 병합 후 검증

```python
# ordes의 order_status에 completed인 것만 병합해서 order_items가 출력?
# 불리언 마스크 
#orders["order_status"] == "completed"
completed_orders = orders[orders["order_status"] == "completed"]
print(completed_orders["order_status"].value_counts())
```

```python
# order_items 에서 completed 인 항목만?
items_with_orders=order_items.merge(
    orders[["order_id", "customer_id","order_date","order_status"]],
    on="order_id", how="left", validate="many_to_one",indicator=True,
    )

items_with_orders.head()
```

```python
# 고객별 평균 주문 건수(완료건 기준)
items_with_orders["order_status"].value_counts()
completed_item_with_orders = items_with_orders[items_with_orders["order_status"]=="completed"]
print(completed_item_with_orders["order_status"].value_counts())
```

```python
# 고객별 완료 주문 건수?
customer_completed_orders=completed_item_with_orders.groupby("customer_id")["order_id"].nunique()

customer_completed_orders.head()
```
